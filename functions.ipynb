```python
# functions

# functions are defined using def, followed by a name and parentheses

def greet():
    print("Hey Shanmuka very good you are practicing!! :)")

greet()
```

```python
# functions can take parameters also 

def greet(name):
    print(f"Hello {name} very good you are practicing!!")

greet("Shanmuka")
```

```python
# functions supports default parameters, variable length arguments(args and *kwargs), 
# and keyword arguments
```

```python
# functions can return values using return

def addition(a,b):
    return a + b

print(addition(2,3))
```

```python
# lambda functions

```

```python
# lambda function to return square of a number


# single argument
pythonsquare = lambda x: x**2

print(pythonsquare(5))

```

```python
#multiple arguments

pythonadd = lambda a,b : a + b

print(pythonadd(2,3))
```

```python
#Common use cases include passing lambdas as arguments for functions like map, filter:
pythonnums = [1,2,3,4]

squaredlist = list(map(lambda x: x**2, pythonnums))

print(squaredlist)
```

```python
#Lambda expressions can contain conditional expressions:

numcheck = lambda x: "Even" if x%2 == 0 else "Odd"

print(numcheck(5))

print(numcheck(6))
```

```python
# print squares of even numbers using lambda

numlist = [1,2,3,4,5,6,7]
print(list(map(lambda x: x**2, [x for x in numlist if x%2 == 0 ])))
```
